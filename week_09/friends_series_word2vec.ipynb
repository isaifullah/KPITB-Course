# Now we will perform word2vec it on Friends series data.

```python
import numpy as np
import pandas as pd
```

```python
import gensim
import os
```

```python
from nltk import sent_tokenize
from gensim.utils import simple_preprocess

# 1. Get the correct path (assuming the file is in the same folder as your script)
file_path = "./Dataset/Friends_series.txt"

# 2. Read the file
with open(file_path, 'r', encoding='utf-8') as f:
    # Read lines and split into words
    story  = [line.lower().split() for line in f if line.strip()]

# Now you can use it in your model
# model.build_vocab(sentences)
```

```python
story
```

```python
model = gensim.models.Word2Vec(
    window=10,
    min_count=2
)
```

```python
model.build_vocab(story)
```

```python
model.train(story, total_examples=model.corpus_count, epochs=model.epochs)
```

```python
model.wv.most_similar('ross')
```

```python
model.wv.doesnt_match(['ross','joy','robb','phoebe','airport','monica'])
```

```python
model.wv.doesnt_match(['joy', 'rachel', 'emily', 'phoebe'])
```
