```python
from ultralytics import YOLO

model = YOLO("yolov8n-cls.pt")

model.train(
    data='weather_dataset',
    epochs=5,
    imgsz=224
)
```

```python
from ultralytics import YOLO
from pathlib import Path

weights_path = Path("runs/classify/train/weights/best.pt")
if weights_path.exists():
    model = YOLO(str(weights_path))
else:
    print(f"Checkpoint not found at {weights_path.resolve()}. Using existing `model` from training cell.")

```

```python
results = model.predict('C:\\Users\\Saif Ullah\\KPITB-Course\\week 6\\test_img.jpg')
```

```python
for r in results:
    print("Predicted class:", r.names[r.probs.top1])
    print("Confidence:", r.probs.top1conf)
```

### untrained model and orediction based on untrained model

```python
from ultralytics import YOLO

# Load pretrained YOLO classification model
model = YOLO("yolov8n-cls.pt")

# Predict image
results = model("test_img.jpg")

# Print predicted class and confidence
for r in results:
    class_name = r.names[r.probs.top1]
    confidence = r.probs.top1conf

    print("Predicted class:", class_name)
    print("Confidence:", float(confidence))
```
