```python
import pandas as pd

df = pd.read_csv('SMSSpamCollection', sep='\t', names=["label", "message"])
```

```python
df.head()
```

```python
df.shape
```

```python
X=list(df['message'])
```

```python
y=list(df['label'])
```

```python
y
```

```python
y = list(pd.get_dummies(y, drop_first=True)['spam'].astype(int))
```

```python
y
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.20, random_state = 0)
```

```python
from transformers import DistilBertTokenizerFast

tokenizer = DistilBertTokenizerFast.from_pretrained('distilbert-base-uncased')

train_encodings = tokenizer(X_train, truncation=True, padding=True)
test_encodings = tokenizer(X_test, truncation=True, padding=True)
```

```python
import torch

class Dataset(torch.utils.data.Dataset):
    def __init__(self, encodings, labels):
        self.encodings = encodings
        self.labels = labels

    def __getitem__(self, idx):
        item = {key: torch.tensor(val[idx]) for key, val in self.encodings.items()}
        item['labels'] = torch.tensor(self.labels[idx])
        return item

    def __len__(self):
        return len(self.labels)
```

```python
train_dataset = Dataset(train_encodings, y_train)
test_dataset = Dataset(test_encodings, y_test)
```

```python
from transformers import DistilBertForSequenceClassification

model = DistilBertForSequenceClassification.from_pretrained(
    'distilbert-base-uncased',
    num_labels=2
)
```

```python
from transformers import TrainingArguments

training_args = TrainingArguments(
    output_dir='./results',
    num_train_epochs=2,
    per_device_train_batch_size=8,
    per_device_eval_batch_size=16,
    warmup_steps=500,
    weight_decay=0.01,
    logging_dir='./logs',
    logging_steps=10,
)
```

```python
from transformers import Trainer

trainer = Trainer(
    model=model,
    args=training_args,
    train_dataset=train_dataset,
    eval_dataset=test_dataset,
)
```

```python
trainer.train()
```

```python
trainer.evaluate(test_dataset)
```

```python
output=trainer.predict(test_dataset)[1]
```

```python
output
```

```python
from sklearn.metrics import confusion_matrix

cm=confusion_matrix(y_test,output)
cm
```

```python
trainer.save_model('senti_model')
```

```python
#task: load the saved model ('senti_model') and do inferencig on some test email.
from transformers import DistilBertForSequenceClassification
model = DistilBertForSequenceClassification.from_pretrained('senti_model')
from transformers import DistilBertTokenizerFast
tokenizer = DistilBertTokenizerFast.from_pretrained('distilbert-base-uncased')
def predict_sentiment(text):
    inputs = tokenizer(text, return_tensors="pt", truncation=True, padding=True)
    outputs = model(**inputs)
    predicted_label = torch.argmax(outputs.logits).item()
    return "spam" if predicted_label == 1 else "ham"
predict_sentiment("Congratulations! You've won a free ticket to the Bahamas. Click here to claim your prize.")
predict_sentiment("Hey, are we still on for lunch tomorrow?")
```
